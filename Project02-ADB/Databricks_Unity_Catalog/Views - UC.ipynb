## Permanent Views
##### the view persists within catalog/schema even after Cluster gets de-attached

```python
%sql

CREATE VIEW managed_catalog.managed_schema.View1 
AS 
SELECT * FROM delta.`abfss://mycontainer@adlspracticeproject02.dfs.core.windows.net/external_table_data` 
WHERE id=1
```

```python
%sql
SELECT * FROM managed_catalog.managed_schema.View1
```

#TEMP Views
###### this view wouldn't persist after cluster de-attached

```python
%sql

CREATE OR REPLACE TEMP VIEW View2
AS 
SELECT * FROM delta.`abfss://mycontainer@adlspracticeproject02.dfs.core.windows.net/external_table_data` 
WHERE id=1

--as temporary view, we don't need to register/define catalog,schema in format
```

```python
%sql
SELECT * FROM View2
```
