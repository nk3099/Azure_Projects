```python
%sql
DROP VOLUME managed_catalog.managed_schema.myExternalVolume
```

```python
%sql

--creating a volume

CREATE EXTERNAL VOLUME managed_catalog.managed_schema.myExternalVolume
LOCATION "abfss://mycontainer@adlspracticeproject02.dfs.core.windows.net/volumes"

--myexternalvolume points to volumes directory(folder) in datalake storage
```

```python
%python
#copying file for volume

dbutils.fs.cp("abfss://mycontainer@adlspracticeproject02.dfs.core.windows.net/source/Customer.csv", "abfss://mycontainer@adlspracticeproject02.dfs.core.windows.net/volumes")
```
