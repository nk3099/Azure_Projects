## Creating Databricks Parameter

```python
dbutils.widgets.text("parameter_filename","")
```

```python
parameter_filename = dbutils.widgets.get("parameter_filename")

display(parameter_filename)
```

## Reading Data using Autoloader
**Note:** Autoloader is built on top of Sparkstreaming

```python
df = spark.readStream.format("cloudFiles") \
    .option("cloudFiles.format","parquet") \
    .option("cloudFiles.schemaLocation", f"abfss://silver@adlspotifyde.dfs.core.windows.net/checkpoint/checkpoint_{parameter_filename}") \
    .load(f"abfss://bronze@adlspotifyde.dfs.core.windows.net/{parameter_filename}/")
```

## Writing Data using Autoloader

```python
df.writeStream.format("delta") \
    .option("outputMode","append") \
    .option("checkpointLocation",f"abfss://silver@adlspotifyde.dfs.core.windows.net/checkpoint/checkpoint_{parameter_filename}") \
    .option("path",f"abfss://silver@adlspotifyde.dfs.core.windows.net/{parameter_filename}/data") \
    .trigger(once=True) \
    .start()
```
